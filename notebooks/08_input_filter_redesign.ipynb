# 08 — Input Filter Redesign: Average-Current-Limit Method

**Project:** USB-C PD → 12 V fan-array controller (LM5176 4-switch buck-boost, 42 W)

**Purpose:** Redesign of the input-power smoothing after PCB v1: use the LM5176's average input-current limit + soft-start behavior as the 'filter', simulated against the measured fan startup profile (02) including USB-C cable inductance/resistance and source current limiting.

**Method:** time-domain converter simulations (fixed and adaptive timestep) of startup and load steps, with cycle-by-cycle and average current limits modeled.

**See also:** 02 (measured load), 05 (capacitor bank), 07 (loop design).

## Startup simulation (fixed timestep)

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parameters ---
V_in_source = 20.0
R_source = 0.1      
C_in = 4*47e-6 + 2*10e-6       
V_out_nom = 12.0
I_load_max = 7.0    
L = 47e-6
C_out = 4*4700e-6 + 9 * 10e-6

# LM5176 Logic
R_isns_in = 0.1e-6 # 20e-3 
V_in_thresh = 0.050 
R_isns_l = 10e-3    
V_l_thresh = 0.080  
I_l_cbc = V_l_thresh / R_isns_l # 4.0A Peak Limit

g_m_ccl = 1e-3          
C_ss = 0.1e-6
I_ss_up = 5e-6
V_ref_fb = 0.8
beta = V_ref_fb / V_out_nom
f_sw = 100000       
T_sw = 1/f_sw

# Simulation Settings
dt = 1e-7           # 100ns for cleaner switching
t_startup = 0.05
t_total = 0.10
steps = int(t_total / dt)

# State Variables
v_in_node = 0#V_in_source
v_out = 0.0
i_l = 0.0
v_ss = 0.0
v_sns_filt = 0.0    
v_i_in_filt = 0.0
r_filt, c_filt = 1000, 100e-9
high_side_on = False
last_clock = -T_sw

# Results
time = np.linspace(0, t_total, steps)
log_v_out = np.zeros(steps)
log_v_in_node = np.zeros(steps)
log_i_l = np.zeros(steps)
log_i_in_sns_avg = np.zeros(steps)
log_v_ss = np.zeros(steps)
log_i_in_source = np.zeros(steps)
log_i_load = np.zeros(steps)
log_i_in_filt = np.zeros(steps)

for i in range(steps):
    t = time[i]
    
    # 1. Load Profile
    i_load = I_load_max if (t > t_startup and (t % (1/160)) < (1/320)) else 0.0
    
    # 2. Valley Target from Control Loop
    v_target = min(v_ss, V_ref_fb) / beta
    # Valley current target (Setpoint for when to allow the next HS turn-on)
    i_valley_target = (v_target - v_out) * 50.0 + (i_load - 2.0) 
    
    # 3. Fixed Frequency / Valley Logic
    # Check if we are at a new clock cycle
    if (t - last_clock) >= T_sw:
        # In Valley Mode, we turn ON at the clock IF current is below valley setpoint
        if i_l <= i_valley_target:
            high_side_on = True
            last_clock = t
            
    # HS stays on for Ton. In LM5176 Buck, Ton is ~ (Vout/Vin)*T_sw
    t_on_target = (max(v_out, 0.1) / max(v_in_node, 1.0)) * T_sw
    
    if high_side_on:
        # Turn OFF if Ton is reached OR if 4A CBC Peak limit is hit
        if (t - last_clock) >= t_on_target or i_l >= I_l_cbc:
            high_side_on = False

    # 4. Inductor Physics
    v_l = (v_in_node - v_out) if high_side_on else -v_out
    i_l += (v_l / L) * dt
    
    # 5. Input Rail & Average Sense
    i_in_sw = i_l if high_side_on else 0.0
    i_in_source = (V_in_source - v_in_node)/R_source
    v_in_node += (((V_in_source - v_in_node)/R_source) - i_in_sw) * (dt/C_in)
    v_i_in_filt += ((i_in_sw * R_isns_in - v_i_in_filt) / (r_filt * c_filt)) * dt
    v_sns_filt += ((i_in_source * R_isns_in - v_sns_filt) / (r_filt * c_filt)) * dt

    # 6. Soft Start & Output Cap
    i_ss_down = g_m_ccl * max(0, v_sns_filt - V_in_thresh)
    v_ss = np.clip(v_ss + ((I_ss_up - i_ss_down) / C_ss) * dt, 0, 1.21)
    v_out += ((i_l - i_load) / C_out) * dt
    
    # Log
    log_v_out[i], log_v_in_node[i], log_i_l[i], log_i_in_sns_avg[i], log_v_ss[i] = v_out, v_in_node, i_l, v_sns_filt/R_isns_in, v_ss
    log_i_in_source[i] = (V_in_source - v_in_node)/R_source
    log_i_load[i] = i_load
    log_i_in_filt[i] = v_i_in_filt / R_isns_in

# --- Plotting ---
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 12), sharex=True)

ax1.plot(time*1000, log_v_out, color='tab:blue', label='V_out')
ax1.plot(time*1000, log_v_in_node/2, color='tab:green', label='V_in/2')
ax1.set_ylabel('Output Voltage (V)')
ax1.grid(True, alpha=0.3)
ax1.set_title("LM5176: Fixed Frequency Valley Current Mode (Buck)")

ax2.plot(time*1000, log_i_l, color='tab:green', alpha=0.4, label='Inductor Current')
ax2.axhline(I_l_cbc, color='red', linestyle='--', label='CBC Peak Limit')
ax2.axhline(V_in_thresh / R_isns_in, color='grey', linestyle='--', label='Average Input Limit')
ax2.plot(time*1000, log_i_in_sns_avg, color='darkorange', label='Avg Sensed I_in')
ax2.plot(time*1000, log_i_in_source, color='blue', label=' I_in_source', alpha = 0.3)
ax2.plot(time*1000, log_i_in_filt, color='red', label='I_in_filt', alpha = 0.5)
ax2.plot(time*1000, log_i_load, color='red', label=' I_load', alpha = 0.3)
ax2.set_ylabel('Current (A)')
ax2.legend(loc='upper right')
ax2.grid(True, alpha=0.3)
ax2.set_ylim([-5,10])
#ax2.set_xlim([24,30])

ax3.plot(time*1000, log_v_ss, color='tab:purple', label='V_ss')
ax3.set_ylabel('Soft Start Voltage (V)')
ax3.set_xlabel('Time (ms)')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Startup simulation with USB-C cable physics

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
%matplotlib inline
plt.close('all')
# --- Parameters ---
V_in_source = 20.0
# USB-C Cable Physics (1-meter typical)
L_cable = 1.0e-6      # 1uH stray inductance
R_cable = 0.15        # Resistance of wires + connectors (150mOhm)

C_in = 4*47e-6 + 2*1e-6       
V_out_nom = 12.0
I_load_max = 7.0    
L = 47e-6
C_out = 4*4700e-6 + 9 * 1e-6

#f_bw = 700 # v2, choose fast response. 1/3 of rhp boost
#f_zc = 6 # 1.5 times fp1boost, about 3x fp1buck
#f_pc2 = 5000 #7 to 10 times f_bw

r_comp = 79500 #2*np.pi*1.31e-3*(280000+20000)/20000*5*0.01/(5/12)*(4.7e-3*4*0.7) *f_bw #79.5k
c_comp = 300e-9#1 / (2*np.pi*f_zc*r_comp) # 220n old, 334n new
c_hf = 400e-12 #1 / (2*np.pi*f_pc2*r_comp) # 400pf new, old150e-9 
c_slope = 330e-12

r_comp = 0.1e3 #100e3
c_comp = 180e-6 #220e-9 
c_hf = 1e-6#150e-9
c_slope = 330e-12

# LM5176 Logic
R_isns_in = 20e-3 # 20e-3 
V_in_thresh = 0.050 
R_isns_l = 10e-3    
V_l_thresh = 0.080  
I_l_cbc = V_l_thresh / R_isns_l # 8.0A Peak Limit
gm_slope = 2e-6
a_cs = 5

g_m_ccl = 1e-3
v_fb_ss_offset = -18e-3
gm_ea = 1.31e-3      
C_ss = 0.5e-6
I_ss_up = 5e-6
V_ref_fb = 0.8
beta = V_ref_fb / V_out_nom
f_sw = 100000       
T_sw = 1/f_sw

# Simulation Settings
dt = 2e-7           # 200ns timestep - keep small for stability
t_total = 0.03         # Total simulation time
steps = int(t_total / dt)
print(f"Steps: {steps:,}")

# --- Manual Fan Load Profile ---
# Fan is PWM-controlled with 50% duty cycle square wave
# - Speed = PWM frequency (Hz)
# - Current = PEAK current during ON phase
# - Average load current = peak * 0.5 (duty cycle)
#
# All values linearly ramp between segment endpoints (except sudden jump at 0.5s)
# Note: Current column shows PEAK values, not average
#
# Time(s) | Speed(Hz) | Current Peak(A) | Description
# --------|-----------|-----------------|-------------
# 0.0     |    0      |       0         | Start
# 0.5     |    0      |       0         | Startup delay
# 0.51    |   10      |       5         | Sudden jump: speed 0→10Hz, current 0→5A
# 1.0     |   60      |       4         | Speed ramps 10→60Hz, current ramps 5→4A
# 1.5     |  110      |       9         | Speed ramps 60→110Hz, current ramps 4→9A
# 2.0     |  150      |       7         | Speed ramps 110→150Hz, current ramps 9→7A
# 2.5     |  160      |       7         | Speed ramps 150→160Hz, current holds at 7A
# 3.0     |  160      |       0         | Speed holds, current ramps 7→0A
# 3.5     |  160      |       0         | Speed holds at 160Hz, current = 0
# 5.0     |    0      |       0         | Speed ramps 160→0Hz, current = 0
#
# Average load = peak * 0.5 (50% duty cycle)

# Define profile as time segments: (end_time, speed_hz, current_peak_amps)
profile_segments = [
    (0.0,   0,    0.0),   # start
    (0.5,   0,    0.0),   # startup delay
    (0.51, 10,    5.0),   # sudden jump: speed 0→10Hz, current 0→5A
    (1.0,  60,    4.0),   # speed: 30→60Hz, current peak: 5→4A
    (1.5, 110,    9.0),   # speed: 60→110Hz, current peak: 4→9A
    (2.0, 150,    7.0),   # speed: 110→150Hz, current peak: 9→7A
    (2.5, 160,    7.0),   # speed: 150→160Hz, current peak: 7A (hold)
    (3.0, 160,    0.0),   # speed: 160Hz (hold), current peak: 7→0A
    (3.5, 160,    0.0),   # speed: 160Hz (hold), current = 0
    (5.0,   0,    0.0),   # speed: 160→0Hz, current = 0
]

# Pre-compute fan profile using vectorized operations (much faster)
# Build arrays for speed and current at high resolution
t_profile = np.linspace(0, t_total, 50000)  # 50k points for smooth profile

# Create speed and current arrays by linear interpolation through segments
speed_profile = np.zeros_like(t_profile)
current_peak_profile = np.zeros_like(t_profile)

for i in range(len(profile_segments) - 1):
    t_start = profile_segments[i][0]
    t_end = profile_segments[i+1][0]
    speed_start = profile_segments[i][1]
    speed_end = profile_segments[i+1][1]
    current_start = profile_segments[i][2]
    current_end = profile_segments[i+1][2]
    
    mask = (t_profile >= t_start) & (t_profile < t_end)
    if np.any(mask):
        frac = (t_profile[mask] - t_start) / (t_end - t_start)
        speed_profile[mask] = speed_start + frac * (speed_end - speed_start)
        current_peak_profile[mask] = current_start + frac * (current_end - current_start)

# Handle last segment
if profile_segments[-1][0] <= t_total:
    last_seg = len(profile_segments) - 1
    mask = t_profile >= profile_segments[last_seg][0]
    speed_profile[mask] = profile_segments[last_seg][1]
    current_peak_profile[mask] = profile_segments[last_seg][2]

# Pre-compute phase accumulation (cumulative integral of speed)
# Use cumulative trapezoidal integration for phase
phase_profile = np.zeros_like(t_profile)
for i in range(1, len(t_profile)):
    dt_phase = t_profile[i] - t_profile[i-1]
    # Use average speed in this interval
    phase_profile[i] = phase_profile[i-1] + 0.5 * (speed_profile[i] + speed_profile[i-1]) * dt_phase

# Apply PWM: ON when fractional phase < 0.5
phase_frac = phase_profile % 1.0
pwm_on = (phase_frac < 0.5) & (speed_profile > 0)
i_profile = np.where(pwm_on, current_peak_profile, 0.0)

# Create interpolator for simulation
f_load_scale = interp1d(t_profile, i_profile, kind='linear', fill_value=0.0)
f_speed_scale = interp1d(t_profile, speed_profile, kind='linear', fill_value=0.0)
f_current_peak_scale = interp1d(t_profile, current_peak_profile, kind='linear', fill_value=0.0)

print(f"Manual fan profile: max peak current = {np.max(i_profile):.2f}A")
print(f"Profile: 50% duty cycle PWM, speed 0-160Hz")

# State Variables
v_in_node = 0#V_in_source
i_cable = 0
v_out = 0.0
i_l = 0.0
v_ss = 0.0
v_sns_filt = 0.0    
v_i_in_filt = 0.0
r_filt, c_filt = 500, 5e-9
high_side_on = False
last_clock = -T_sw
# --- New State Variables for COMP Pin ---
v_comp = 1.0        # The actual voltage on the COMP pin (Top of C_hf)
v_c_comp = 1.0      # The voltage across the "bulk" capacitor C_comp only
v_fb = 0.0      # Feedback voltage
i_comp = 0.0    # Current out of Gm amplifier
cbc_limit_flag = False

# PID Variables (persist across calls)
error_prior = 0
integral = 0
Kp = 1e-9 #0.02  # Start small
Ki = 1e-9# 50000 # Slow integration
Kd = 1e-9# 0.0000004 # Minimal derivative to dampen
def update_ss_control(target_i, actual_i, dt):
    global error_prior, integral
    global log_integral, i, log_kp, log_kd
    error = target_i - actual_i
    
    # Integral with anti-windup (only integrate if not saturated)
    integral += error * dt
    log_integral[i] = integral * Ki
    # Derivative (helps with fast spikes)
    derivative = (error - error_prior) / dt
    log_kd[i] = Kd * derivative
    log_kp[i] = Kp * error
    
    # Calculate output
    output = (Kp * error) + (Ki * integral) + (Kd * derivative)
    
    # Map output to the SS voltage range (e.g., 0.8V offset)
    v_ss_drive = np.clip(0.782 + output, 0.0, 1.2)
    if v_ss_drive > 0.782:
        log_integral[i] = log_integral[i-1]
        integral = log_integral[i-1] / Ki
    
    error_prior = error
    return v_ss_drive

# Results
time = np.linspace(0, t_total, steps)
log_v_out = np.zeros(steps)
log_v_in_node = np.zeros(steps)
log_i_l = np.zeros(steps)
log_i_in_sns_avg = np.zeros(steps)
log_v_ss = np.zeros(steps)
log_i_in_source = np.zeros(steps)
log_i_load = np.zeros(steps)
log_i_cable = np.zeros(steps)
log_i_in_filt = np.zeros(steps)
log_v_comp = np.zeros(steps)
log_v_c_comp = np.zeros(steps)
log_i_valley_target_compensated = np.zeros(steps)
log_i_ss_down = np.zeros(steps)
log_integral = np.zeros(steps)
log_kd = np.zeros(steps)
log_kp = np.zeros(steps)
log_gate = np.zeros(steps)
log_fan_speed = np.zeros(steps)  # Track fan speed (Hz)
log_fan_current_peak = np.zeros(steps)  # Track fan peak current (A)

for i in range(steps):
    t = time[i]
    
    # Log fan speed and peak current (use pre-computed interpolators - much faster)
    log_fan_speed[i] = f_speed_scale(t)
    log_fan_current_peak[i] = f_current_peak_scale(t)
    
    # 1. Load Profile - using manual fan profile with PWM
    i_load = f_load_scale(t) if t <= t_total else 0.01

    # 2. FEEDBACK & ERROR AMP LOGIC (The "Pro" Way)
    # The FB pin sees V_out through the resistor divider
    v_fb = v_out * beta 
    
    # Current into COMP pin: i = gm * (Vref - Vfb)
    # Limited by the 280uA source/sink capability you provided
    i_err_amp_raw = gm_ea * (min(v_ss-v_fb_ss_offset, V_ref_fb) - v_fb)
    i_err_amp = np.clip(i_err_amp_raw, -280e-6, 280e-6)
    
    # 1. Calculate current flowing through the Resistor branch
    # This depends on the difference between the COMP pin and the bulk capacitor
    i_r_branch = (v_comp - v_c_comp) / r_comp

    # 2. Update the COMP pin voltage
    # The net current into the COMP node is (i_err_amp - i_r_branch).
    # This current charges/discharges C_hf.
    dv_comp = ((i_err_amp - i_r_branch) / c_hf) * dt
    v_comp += dv_comp

    # 3. Update the Bulk Capacitor voltage
    # The current that went through the resistor now charges C_comp.
    dv_c_comp = (i_r_branch / c_comp) * dt
    v_c_comp += dv_c_comp

    # 4. Limit v_comp to the chip's physical rails (usually 0V to ~3.3V)
    v_comp = np.clip(v_comp, 0.3, 3.0)
    
    # 4. TRANSLATE COMP TO CURRENT TARGET
    # In most controllers, I_target = (V_comp - offset) * Gain
    # Adjust 1.2 and 5.0 to match your specific IC's 
    i_valley_target = (v_comp - 1.6) / a_cs / R_isns_l

    # --- Real LM5176 Deadbeat Physics ---
    # The IC senses Vout to set the falling-edge compensation in Buck mode
    i_slope_internal = gm_slope * (v_in_node - v_out) 
    # Voltage slope on the COMP pin
    dv_slope_dt = i_slope_internal / c_slope 
    # Effective Amps slope (what the comparator 'thinks' is happening)
    se_eff = dv_slope_dt / (R_isns_l * a_cs)

    # Update the target
    t_cycle = (t - last_clock)
    slope_ramp = se_eff * t_cycle
    i_valley_target_compensated = min(I_l_cbc, i_valley_target + slope_ramp)
    log_i_valley_target_compensated[i] = i_valley_target_compensated

    
    # 3. Fixed Frequency / Valley Logic
    # Check if we are at a new clock cycle
    if i_l <= i_valley_target_compensated and not(cbc_limit_flag):
        high_side_on = True
    #if i_l >= I_l_cbc:
    ## Turn OFF if Ton is reached OR if 4A CBC Peak limit is hit
    #    cbc_limit_flag = True
    #    high_side_on = False
    if high_side_on or cbc_limit_flag:
        log_i_valley_target_compensated[i] = log_i_valley_target_compensated[i-1]
    if (t - last_clock) >= T_sw:
        cbc_limit_flag = False
        if i_l>I_l_cbc:# high_side_on == False: #never turned on
            cbc_limit_flag = True
        high_side_on = False
        # In Valley Mode, we turn ON at the clock IF current is below valley setpoint
        last_clock = t
    
        
    # HS stays on for Ton. In LM5176 Buck, Ton is ~ (Vout/Vin)*T_sw
    #t_on_target = (max(v_out, 0.1) / max(v_in_node, 1.0)) * T_sw


    # 4. Inductor Physics
    v_l = (v_in_node - v_out) if high_side_on else -v_out
    i_l += (v_l / L) * dt
    # 5. USB-C CABLE & INPUT RAIL PHYSICS (The LC Pi Filter)
    # Cable current: di/dt = (V_source - V_drop_R - V_pcb) / L_cable
    di_cable = (V_in_source - (i_cable * R_cable) - v_in_node) / L_cable
    i_cable += di_cable * dt
    
    # 5. Input Rail & Average Sense
    i_in_sw = i_l if high_side_on else 0.0
    #i_in_source = (V_in_source - v_in_node)/R_source
    #v_in_node += (((V_in_source - v_in_node)/R_source) - i_in_sw) * (dt/C_in)
    
    v_in_node += (i_cable - i_in_sw) * (dt / C_in)
    #v_i_in_filt += ((i_in_sw * R_isns_in - v_i_in_filt) / (r_filt * c_filt)) * dt
    v_sns_filt += ((i_cable * R_isns_in - v_sns_filt) / (r_filt * c_filt)) * dt

    # 1. Load Profile - using manual fan profile
    i_load = f_load_scale(t)

    # 6. Soft Start & Output Cap
    i_ss_down = np.clip(g_m_ccl * (v_sns_filt - V_in_thresh), -20e-6, 100e-6) #allow -20 pulldown, not realistic for lm5176, but maybe with external circuit
    i_ss_down = np.clip(g_m_ccl * (v_sns_filt - V_in_thresh), 0, 100e-6) #allow -20 pulldown, not realistic for lm5176, but maybe with external circuit
    v_ss = np.clip(v_ss + ((I_ss_up - i_ss_down) / C_ss) * dt, 0, 0.85) #1.21 from datasheet but test cap manually externally
    # v_ss = np.clip((V_in_thresh - v_sns_filt)*1+0.782, 0, 0.85) #1.21 from datasheet but test cap manually externally
    # v_ss = update_ss_control(3, v_sns_filt/R_isns_in, dt)
    v_out += ((i_l - i_load) / C_out) * dt
    
    # Log
    log_v_out[i], log_v_in_node[i], log_i_l[i], log_i_in_sns_avg[i], log_v_ss[i] = v_out, v_in_node, i_l, v_sns_filt/R_isns_in, v_ss
    #log_i_in_source[i] = (V_in_source - v_in_node)/R_cable
    log_i_load[i] = i_load
    log_i_cable[i] = i_cable
    log_i_in_filt[i] = v_i_in_filt / R_isns_in
    log_v_comp[i] = v_comp
    log_v_c_comp[i] = v_c_comp
    log_i_ss_down[i] = i_ss_down
    log_gate[i] = high_side_on

# --- Plotting ---
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(10, 14), sharex=True)

ax1.plot(time*1000, log_v_out, color='tab:blue', label='V_out')
ax1.plot(time*1000, log_v_in_node/2, color='tab:green', label='V_in/2')
ax1.set_ylabel('Output Voltage (V)')
ax1.grid(True, alpha=0.3)
ax1.set_title("LM5176: Fixed Frequency Valley Current Mode (Buck)")

ax2.plot(time*1000, log_i_l, color='tab:green', alpha=0.4, label='Inductor Current')
ax2.axhline(I_l_cbc, color='red', linestyle='--', label='CBC Peak Limit')
ax2.axhline(V_in_thresh / R_isns_in, color='grey', linestyle='--', label='Average Input Limit')
ax2.plot(time*1000, log_i_in_sns_avg, color='darkorange', label='Avg Sensed I_in')
#ax2.plot(time*1000, log_i_in_source, color='blue', label=' I_in_source', alpha = 0.3)
ax2.plot(time*1000, log_i_cable, color='blue', label=' I_cable', alpha = 0.3)
#ax2.plot(time*1000, log_i_in_filt, color='red', label='I_in_filt', alpha = 0.5)
ax2.plot(time*1000, log_i_load, color='red', label=' I_load', alpha = 0.3)
ax2.plot(time*1000, (log_v_comp - 1.6) / a_cs / R_isns_l, color='yellow', label='valley target', alpha = 0.8)
ax2.plot(time*1000, log_i_valley_target_compensated, color='purple', label='valley target compensated', alpha = 0.3)
ax2twin = ax2.twinx()
ax2twin.plot(time*1000, log_v_comp, color='lightblue', label='v_comp', alpha = 0.6)
ax2twin.plot(time*1000, log_v_c_comp, color='darkblue', label='v_c_comp', alpha =0.6)
ax2twin.plot(time*1000, log_gate/10, color='red', linestyle='--', label='mosfet gate', alpha = 0.1)
ax2.set_ylabel('Current (A)')
ax2.legend(loc='upper left')
ax2twin.legend(loc='upper right')
ax2.grid(True, alpha=0.3)
ax2.set_ylim([-5,30])
#ax2.set_xlim([30,33])

ax3.plot(time*1000, log_v_ss, color='tab:purple', label='V_ss')
ax3twin = ax3.twinx()
#ax3twin.plot(time*1000, log_i_ss_down, color='blue', label=' I_ss_down', alpha = 0.8)
ax3twin.plot(time*1000, log_kp, color='blue', label=' kp', alpha = 0.8)
ax3twin.plot(time*1000, log_integral, color='red', label=' ki', alpha = 0.8)
ax3twin.plot(time*1000, log_kd, color='orange', label=' kd', alpha = 0.8)
ax3twin.legend()
ax3.set_ylabel('Soft Start Voltage (V)')
ax3twin.set_ylim([-0.2,0.2])
ax3.grid(True, alpha=0.3)

# Fan speed and current profile
ax4.plot(time*1000, log_fan_speed, color='tab:blue', label='Fan Speed (Hz)')
ax4.set_ylabel('Speed (Hz)', color='tab:blue')
ax4.tick_params(axis='y', labelcolor='tab:blue')
ax4twin = ax4.twinx()
ax4twin.plot(time*1000, log_fan_current_peak, color='tab:red', linestyle='--', label='Peak Current (A)')
ax4twin.plot(time*1000, log_i_load, color='tab:red', alpha=0.5, label='Instantaneous Load (A)')
ax4twin.set_ylabel('Current (A)', color='tab:red')
ax4twin.tick_params(axis='y', labelcolor='tab:red')
ax4.set_xlabel('Time (ms)')
ax4.grid(True, alpha=0.3)
ax4.legend(loc='upper left')
ax4twin.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
# Adaptive Timestep Simulation for Buck-Boost Converter
# Instead of fixed 200ns steps, we calculate when events happen and jump to them

import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
%matplotlib inline
plt.close('all')

# --- Parameters ---
V_in_source = 20.0
L_cable = 1.0e-6
R_cable = 0.15

C_in = 4*47e-6 + 2*1e-6       
V_out_nom = 12.0
I_load_max = 7.0    
L = 47e-6
C_out = 4*4700e-6 + 9 * 1e-6

r_comp = 2e3 #100e3
c_comp = 47e-6 #220e-9 # if put 0805 on 0603 pad, or stack
c_hf = 150e-9#150e-9
c_slope = 330e-12

R_isns_in = 10e-3 
V_in_thresh = 0.050 
R_isns_l = 10e-3    
V_l_thresh = 0.080  
I_l_cbc = V_l_thresh / R_isns_l
gm_slope = 2e-6
a_cs = 5

g_m_ccl = 1e-3
v_fb_ss_offset = -18e-3
gm_ea = 1.31e-3      
C_ss = 0.5e-6
I_ss_up = 5e-6
V_ref_fb = 0.8
beta = V_ref_fb / V_out_nom
f_sw = 100000       
T_sw = 1/f_sw

# --- Fan Profile (same as before) ---
profile_segments = [
    (0.0,   0,    0.0),
    (0.8/4,   0,    0.0),
    (0.81/4, 10,    5.0),
    (1.2/4,  60,    4.0),
    (1.7/4, 110,    9.0),
    (2.2/4, 150,    7.0),
    (2.7/4, 160,    7.0),
    (3.2/4, 160,    0.0),
    (3.7/4, 160,    0.0),
    (5.2/4,   0,    0.0),
]

t_profile = np.linspace(0, 5, 50000)
speed_profile = np.zeros_like(t_profile)
current_peak_profile = np.zeros_like(t_profile)

for i in range(len(profile_segments) - 1):
    t_start, speed_start, current_start = profile_segments[i]
    t_end, speed_end, current_end = profile_segments[i+1]
    mask = (t_profile >= t_start) & (t_profile < t_end)
    if np.any(mask):
        frac = (t_profile[mask] - t_start) / (t_end - t_start)
        speed_profile[mask] = speed_start + frac * (speed_end - speed_start)
        current_peak_profile[mask] = current_start + frac * (current_end - current_start)

# PWM phase
phase_profile = np.zeros_like(t_profile)
for i in range(1, len(t_profile)):
    phase_profile[i] = phase_profile[i-1] + 0.5 * (speed_profile[i] + speed_profile[i-1]) * (t_profile[i] - t_profile[i-1])

phase_frac = phase_profile % 1.0
pwm_on = (phase_frac < 0.5) & (speed_profile > 0)
i_profile = np.where(pwm_on, current_peak_profile, 0.0)

f_load_scale = interp1d(t_profile, i_profile, kind='linear', fill_value=0.0)
f_speed_scale = interp1d(t_profile, speed_profile, kind='linear', fill_value=0.0)
f_current_peak_scale = interp1d(t_profile, current_peak_profile, kind='linear', fill_value=0.0)

# --- Adaptive Timestep Simulation ---
t_total = 1
dt_slow = 1.8e-6    # 2µs - during quiet periods
dt_mid = 600e-9     # 500ns - moderate events
dt_fast = 150e-9  # 200ns - near events
#margin = 400e-9     # 1µs before event to switch to fast mode

# State variables
t = 0.0
v_in_node = 0.0
i_cable = 0.0
v_out = 0.0
i_l = 0.0
v_ss = 0.0
v_sns_filt = 0.0    
r_filt, c_filt = 500, 5e-9 # Filter for sensed current with cutoff frequency of about 63kHz
high_side_on = False
last_clock = 0
v_comp = 1.0        
v_c_comp = 1.0      
v_fb = 0.0      
cbc_limit_flag = False
next_clock = T_sw  # Next clock edge time

# Results storage
time = [0]
log_v_out = [0]
log_v_in_node = [0]
log_i_l = [0]
log_i_in_sns_avg = [0]
log_v_ss = [0]
log_i_load = [0]
log_i_cable = [0]
log_v_comp = [1.0]
log_v_c_comp = [1.0]
log_gate = [False]
log_fan_speed = [0]
log_fan_current_peak = [0]
log_i_valley_target = [(v_comp - 1.6)/(a_cs * R_isns_l)]
log_i_valley_target_compensated = [0]
log_dt_to_valley = []
log_dt_to_clock = []
# Main simulation loop
print("Starting adaptive timestep simulation...")
iter_count = 0
max_iter = 5e6

while t < t_total and iter_count < max_iter:
    iter_count += 1
    
    # Get current values
    i_load = f_load_scale(t)
    fan_speed = f_speed_scale(t)
    fan_current_peak = f_current_peak_scale(t)
    
    # Feedback & error amp (use current v_comp to estimate valley target)
    v_fb = v_out * beta 
    i_err_amp_raw = gm_ea * (min(v_ss-v_fb_ss_offset, V_ref_fb) - v_fb)
    i_err_amp = np.clip(i_err_amp_raw, -280e-6, 280e-6)
    
    # Current valley target (based on previous COMP state)
    i_valley_target = (v_comp - 1.6) / a_cs / R_isns_l
    # Slope compensation
    i_slope_internal = gm_slope * (v_in_node - v_out) 
    se_eff = i_slope_internal / (c_slope * R_isns_l * a_cs)
    i_valley_target_compensated = min(I_l_cbc, i_valley_target + se_eff * (t - last_clock))

    # Time to valley (estimate)
    v_l_off = -v_out
    di_dt_off = v_l_off / L if abs(v_l_off) > 0.1 else 0
    #i_valley_target_compensated_next_large_step = min(I_l_cbc, i_valley_target + se_eff * (t_cycle+dt_slow))
    #dt_to_valley_large_step = (i_l - i_valley_target_compensated_next_large_step) / abs(di_dt_off) if di_dt_off < 0 and abs(di_dt_off) > 1 else dt_slow
    denominator = se_eff - di_dt_off

    if denominator > 0: # Ensure they are actually converging
        dt_to_valley = (i_l - (i_valley_target + se_eff * (t-last_clock))) / denominator
    else:
        dt_to_valley = T_sw # No collision imminent
    dt_to_valley = np.min([dt_to_valley, T_sw])

    # Choose dt: use fast steps near events
    dt = dt_slow
    if not(high_side_on) and not(cbc_limit_flag):
        # if dt_to_valley < dt_slow or (next_clock - t < dt_slow):
        #    dt = dt_mid
        #    if dt_to_valley < dt_mid or (next_clock - t < dt_mid):
        #        dt = dt_fast
        # using 3e-6 as max step cause -0.2 amp "offset" input current when off..
        # using 2e-6 as max step caused -0.1A "offset" current when off..bad..
        dt = np.clip(dt_to_valley*0.7, dt_fast, 2e-6) # 2nd method: scale down as we approach valley to avoid overshoot
    if high_side_on or cbc_limit_flag:
        if (next_clock - t) < dt_slow:
            dt = dt_mid
            if (next_clock - t) < dt_mid:
                dt = dt_fast
        dt = np.clip((next_clock - t)*0.7, dt_fast, 2e-6)
    #dt = min(dt, dt_slow)
    log_dt_to_valley.append(dt_to_valley)
    log_dt_to_clock.append(next_clock - t)
    

    # --- Check for switching events at this timestep ---



        # 3. Fixed Frequency / Valley Logic
    # Check if we are at a new clock cycle
    # print(i_l, i_valley_target_compensated, cbc_limit_flag)
    if i_l <= i_valley_target_compensated and not(cbc_limit_flag):
        high_side_on = True
    #if i_l >= I_l_cbc:
    ## Turn OFF if Ton is reached OR if 4A CBC Peak limit is hit
    #    cbc_limit_flag = True
    #    high_side_on = False
    #if high_side_on or cbc_limit_flag:
    #    log_i_valley_target_compensated[i] = log_i_valley_target_compensated[i-1]
    if t >= next_clock:
        cbc_limit_flag = False
        if i_l>I_l_cbc:# high_side_on == False: #never turned on
            cbc_limit_flag = True
        high_side_on = False
        # In Valley Mode, we turn ON at the clock IF current is below valley setpoint
        last_clock += T_sw
        next_clock += T_sw

    
    t_cycle = t + dt - last_clock
    if (high_side_on == False):
        i_valley_target_compensated = min(I_l_cbc, i_valley_target + se_eff * t_cycle)
    else:
        i_valley_target_compensated = i_valley_target_compensated #hold during on time

    # NOW update COMP voltages with the chosen dt
    i_r_branch = (v_comp - v_c_comp) / r_comp
    v_comp += (i_err_amp - i_r_branch) / c_hf * dt
    v_c_comp += i_r_branch / c_comp * dt
    v_comp = np.clip(v_comp, 0.3, 3.0)
    
    # Inductor
    v_l = (v_in_node - v_out) if high_side_on else -v_out
    i_l += (v_l / L) * dt
    v_l = (v_in_node - v_out) if high_side_on else -v_out
    
    # Cable & input
    di_cable = (V_in_source - (i_cable * R_cable) - v_in_node) / L_cable
    i_in_sw = i_l if high_side_on else 0.0
    v_in_node += (i_cable - i_in_sw) * (dt / C_in)
    v_sns_filt += ((i_cable * R_isns_in - v_sns_filt) / (r_filt * c_filt)) * dt
    i_cable += di_cable * dt
    i_in_sw = i_l if high_side_on else 0.0
    
    
    
    # Soft start & output
    i_ss_down = g_m_ccl * max(0, v_sns_filt - V_in_thresh)
    v_ss = np.clip(v_ss + ((I_ss_up - i_ss_down) / C_ss) * dt, 0, 0.85)
    v_out += ((i_l - i_load) / C_out) * dt
    
    # Time advanced during this iteration
    t += dt
    # Log
    time.append(t)
    log_v_out.append(v_out)
    log_v_in_node.append(v_in_node)
    log_i_l.append(i_l)
    log_i_in_sns_avg.append(v_sns_filt/R_isns_in)
    log_v_ss.append(v_ss)
    log_i_load.append(i_load)
    log_i_cable.append(i_cable)
    log_v_comp.append(v_comp)
    log_v_c_comp.append(v_c_comp)
    log_gate.append(1.0 if high_side_on else 0.0)
    log_fan_speed.append(fan_speed)
    log_fan_current_peak.append(fan_current_peak)
    log_i_valley_target_compensated.append(i_valley_target_compensated)
    


print(f"Simulation complete: {iter_count} steps vs {int(t/200e-9):,} fixed-step iterations")
print(f"Speedup: {int(t/200e-9)/iter_count:.1f}x")

# Convert to arrays
time = np.array(time)
log_v_out = np.array(log_v_out)
log_i_l = np.array(log_i_l)
log_i_load = np.array(log_i_load)
log_v_comp = np.array(log_v_comp)
log_i_valley_target_compensated = np.array(log_i_valley_target_compensated)
log_gate = np.array(log_gate)
log_fan_speed = np.array(log_fan_speed)
log_fan_current_peak = np.array(log_fan_current_peak)
log_v_in_node = np.array(log_v_in_node)
log_dt_to_valley = np.array(log_dt_to_valley)
log_dt_to_clock = np.array(log_dt_to_clock)

# --- Plotting ---
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(10, 14), sharex=True)

ax1.plot(time*1000, log_v_out, color='tab:blue', label='V_out')
ax1.plot(time*1000, log_v_in_node/2, color='tab:green', label='V_in/2')
ax1.set_ylabel('Output Voltage (V)')
ax1.grid(True, alpha=0.3)
ax1.set_title("LM5176: Fixed Frequency Valley Current Mode (Buck)")

ax2.plot(time*1000, log_i_l, color='tab:green', marker = None, alpha=0.4, label='Inductor Current')
ax2.axhline(I_l_cbc, color='red', linestyle='--', label='CBC Peak Limit')
ax2.axhline(V_in_thresh / R_isns_in, color='grey', linestyle='--', label='Average Input Limit')
ax2.plot(time*1000, log_i_in_sns_avg, color='darkorange', label='Avg Sensed I_in')
#ax2.plot(time*1000, log_i_in_source, color='blue', label=' I_in_source', alpha = 0.3)
ax2.plot(time*1000, log_i_cable, color='blue', label=' I_cable', alpha = 0.3)
#ax2.plot(time*1000, log_i_in_filt, color='red', label='I_in_filt', alpha = 0.5)
ax2.plot(time*1000, log_i_load, color='red', label=' I_load', alpha = 0.3)
ax2.plot(time*1000, (log_v_comp - 1.6) / a_cs / R_isns_l, color='yellow', label='valley target', alpha = 0.8)
ax2.plot(time*1000, log_i_valley_target_compensated, color='purple', label='valley target compensated', alpha = 0.2)
ax2twin = ax2.twinx()
ax2twin.plot(time*1000, log_v_comp, color='lightblue', label='v_comp', alpha = 0.6)
ax2twin.plot(time*1000, log_v_c_comp, color='darkblue', label='v_c_comp', alpha =0.6)
ax2twin.plot(time*1000, log_gate/10, color='red', linestyle='--', label='mosfet gate', alpha = 0.1)
ax2.set_ylabel('Current (A)')
ax2.legend(loc='upper left')
ax2twin.legend(loc='upper right')
ax2.grid(True, alpha=0.3)
ax2.set_ylim([-5,15])
#ax2.set_xlim([30,33])

ax3.plot(time*1000, log_v_ss, color='tab:purple', label='V_ss')
ax3twin = ax3.twinx()
#ax3twin.plot(time*1000, log_i_ss_down, color='blue', label=' I_ss_down', alpha = 0.8)
#ax3twin.plot(time*1000, log_kp, color='blue', label=' kp', alpha = 0.8)
#ax3twin.plot(time*1000, log_integral, color='red', label=' ki', alpha = 0.8)
#ax3twin.plot(time*1000, log_kd, color='orange', label=' kd', alpha = 0.8)
#ax3twin.plot(time[:-1]*1000, log_dt_to_valley*1e6, color='blue', label='dt to valley (µs)', alpha = 0.8  )
#ax3twin.plot(time[:-1]*1000, log_dt_to_clock*1e6, color='green', label='dt to clock (µs)', alpha = 0.8  )
ax3twin.legend()
ax3.set_ylabel('Soft Start Voltage (V)')
#ax3twin.set_ylim([-0.2,0.2])
ax3.grid(True, alpha=0.3)
```

```python
# Adaptive Timestep Simulation for Buck-Boost Converter
# Instead of fixed 200ns steps, we calculate when events happen and jump to them

import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
%matplotlib inline
plt.close('all')

# --- Parameters ---
V_in_source = 20.0
L_cable = 1.0e-6
R_cable = 0.15

C_in = 4*47e-6 + 2*1e-6       
V_out_nom = 12.0
I_load_max = 7.0    
L = 47e-6
C_out = 4*4700e-6 + 9 * 1e-6

r_comp = 2e3 #100e3
c_comp = 10e-6 #220e-9 # if put 0805 on 0603 pad, or stack
c_hf = 150e-9#150e-9
c_slope = 330e-12

R_isns_in = 10e-3 
V_in_thresh = 0.050 
R_isns_l = 10e-3    
V_l_thresh = 0.080  
I_l_cbc = V_l_thresh / R_isns_l
gm_slope = 2e-6
a_cs = 5

g_m_ccl = 1e-3
v_fb_ss_offset = -18e-3
gm_ea = 1.31e-3      
C_ss = 0.5e-6
I_ss_up = 5e-6
V_ref_fb = 0.8
beta = V_ref_fb / V_out_nom
f_sw = 100000       
T_sw = 1/f_sw

# --- Fan Profile (same as before) ---
profile_segments = [
    (0.0,   0,    0.0),
    (0.8/4,   0,    0.0),
    (0.81/4, 10,    5.0),
    (1.2/4,  60,    4.0),
    (1.7/4, 110,    9.0),
    (2.2/4, 150,    7.0),
    (2.7/4, 160,    7.0),
    (3.2/4, 160,    0.0),
    (3.7/4, 160,    0.0),
    (5.2/4,   0,    0.0),
]

t_profile = np.linspace(0, 5, 50000)
speed_profile = np.zeros_like(t_profile)
current_peak_profile = np.zeros_like(t_profile)

for i in range(len(profile_segments) - 1):
    t_start, speed_start, current_start = profile_segments[i]
    t_end, speed_end, current_end = profile_segments[i+1]
    mask = (t_profile >= t_start) & (t_profile < t_end)
    if np.any(mask):
        frac = (t_profile[mask] - t_start) / (t_end - t_start)
        speed_profile[mask] = speed_start + frac * (speed_end - speed_start)
        current_peak_profile[mask] = current_start + frac * (current_end - current_start)

# PWM phase
phase_profile = np.zeros_like(t_profile)
for i in range(1, len(t_profile)):
    phase_profile[i] = phase_profile[i-1] + 0.5 * (speed_profile[i] + speed_profile[i-1]) * (t_profile[i] - t_profile[i-1])

phase_frac = phase_profile % 1.0
pwm_on = (phase_frac < 0.5) & (speed_profile > 0)
i_profile = np.where(pwm_on, current_peak_profile, 0.0)

f_load_scale = interp1d(t_profile, i_profile, kind='linear', fill_value=0.0)
f_speed_scale = interp1d(t_profile, speed_profile, kind='linear', fill_value=0.0)
f_current_peak_scale = interp1d(t_profile, current_peak_profile, kind='linear', fill_value=0.0)

# --- Adaptive Timestep Simulation ---
t_total = 1
dt_slow = 1e-6#1.8e-6    # 2µs - during quiet periods
dt_mid = 600e-9     # 500ns - moderate events
dt_fast = 150e-9  # 200ns - near events
#margin = 400e-9     # 1µs before event to switch to fast mode

# State variables
t = 0.0
v_in_node = 4.5
i_cable = 0.0
v_out = 11.0
i_l = 0.0
v_ss = max(0, v_fb-0.3) # temp assume do some sort of clamp to prevent too much windup
v_sns_filt = 0.0    
r_filt, c_filt = 500, 5e-9 # Filter for sensed current with cutoff frequency of about 63kHz
high_side_on = False
last_clock = 0
v_comp = 0        
v_c_comp = 0      
v_fb = v_out * beta      
cbc_limit_flag = False
next_clock = T_sw  # Next clock edge time

# Results storage
time = [0]
log_v_out = [v_out]
log_v_in_node = [v_in_node]
log_i_l = [0]
log_i_in_sns_avg = [0]
log_v_ss = [v_ss]
log_i_load = [0]
log_i_cable = [0]
log_v_comp = [v_comp]
log_v_c_comp = [v_c_comp]
log_gate = [False]
log_fan_speed = [0]
log_fan_current_peak = [0]
log_i_valley_target = [(v_comp - 1.6)/(a_cs * R_isns_l)]
log_i_valley_target_compensated = [0]
log_dt_to_valley = []
log_dt_to_clock = []
# Main simulation loop
print("Starting adaptive timestep simulation...")
iter_count = 0
max_iter = 1e5

while t < t_total and iter_count < max_iter:
    iter_count += 1
    
    # Get current values
    i_load = f_load_scale(t)
    fan_speed = f_speed_scale(t)
    fan_current_peak = f_current_peak_scale(t)
    
    # Feedback & error amp (use current v_comp to estimate valley target)
    v_fb = v_out * beta 
    i_err_amp_raw = gm_ea * (min(v_ss-v_fb_ss_offset, V_ref_fb) - v_fb)
    i_err_amp = np.clip(i_err_amp_raw, -280e-6, 280e-6)
    
    # Current valley target (based on previous COMP state)
    i_valley_target = (v_comp - 1.6) / a_cs / R_isns_l
    # Slope compensation
    i_slope_internal = gm_slope * (v_in_node - v_out) 
    se_eff = i_slope_internal / (c_slope * R_isns_l * a_cs)
    i_valley_target_compensated = min(I_l_cbc, i_valley_target + se_eff * (t - last_clock))

    # Time to valley (estimate)
    v_l_off = -v_out
    di_dt_off = v_l_off / L if abs(v_l_off) > 0.1 else 0
    #i_valley_target_compensated_next_large_step = min(I_l_cbc, i_valley_target + se_eff * (t_cycle+dt_slow))
    #dt_to_valley_large_step = (i_l - i_valley_target_compensated_next_large_step) / abs(di_dt_off) if di_dt_off < 0 and abs(di_dt_off) > 1 else dt_slow
    denominator = se_eff - di_dt_off

    if denominator > 0: # Ensure they are actually converging
        dt_to_valley = (i_l - (i_valley_target + se_eff * (t-last_clock))) / denominator
    else:
        dt_to_valley = T_sw # No collision imminent
    dt_to_valley = np.min([dt_to_valley, T_sw])

    # Choose dt: use fast steps near events
    dt = dt_slow
    if not(high_side_on) and not(cbc_limit_flag):
        # if dt_to_valley < dt_slow or (next_clock - t < dt_slow):
        #    dt = dt_mid
        #    if dt_to_valley < dt_mid or (next_clock - t < dt_mid):
        #        dt = dt_fast
        # using 3e-6 as max step cause -0.2 amp "offset" input current when off..
        # using 2e-6 as max step caused -0.1A "offset" current when off..bad..
        dt = np.clip(dt_to_valley*0.7, dt_fast, 2e-6) # 2nd method: scale down as we approach valley to avoid overshoot
    if high_side_on or cbc_limit_flag:
        if (next_clock - t) < dt_slow:
            dt = dt_mid
            if (next_clock - t) < dt_mid:
                dt = dt_fast
        dt = np.clip((next_clock - t)*0.7, dt_fast, 2e-6)
    #dt = min(dt, dt_slow)
    log_dt_to_valley.append(dt_to_valley)
    log_dt_to_clock.append(next_clock - t)
    

    # --- Check for switching events at this timestep ---



        # 3. Fixed Frequency / Valley Logic
    # Check if we are at a new clock cycle
    # print(i_l, i_valley_target_compensated, cbc_limit_flag)
    if i_l <= i_valley_target_compensated and not(cbc_limit_flag):
        high_side_on = True
    #if i_l >= I_l_cbc:
    ## Turn OFF if Ton is reached OR if 4A CBC Peak limit is hit
    #    cbc_limit_flag = True
    #    high_side_on = False
    #if high_side_on or cbc_limit_flag:
    #    log_i_valley_target_compensated[i] = log_i_valley_target_compensated[i-1]
    if t >= next_clock:
        cbc_limit_flag = False
        if i_l>I_l_cbc:# high_side_on == False: #never turned on
            cbc_limit_flag = True
        high_side_on = False
        # In Valley Mode, we turn ON at the clock IF current is below valley setpoint
        last_clock += T_sw
        next_clock += T_sw

    
    t_cycle = t + dt - last_clock
    if (high_side_on == False):
        i_valley_target_compensated = min(I_l_cbc, i_valley_target + se_eff * t_cycle)
    else:
        i_valley_target_compensated = i_valley_target_compensated #hold during on time

    # NOW update COMP voltages with the chosen dt
    i_r_branch = (v_comp - v_c_comp) / r_comp
    v_comp += (i_err_amp - i_r_branch) / c_hf * dt
    v_c_comp += i_r_branch / c_comp * dt
    v_comp = np.clip(v_comp, 0.3, 3.0)
    
    # Inductor
    v_l = (v_in_node - v_out) if high_side_on else -v_out
    i_l += (v_l / L) * dt
    v_l = (v_in_node - v_out) if high_side_on else -v_out
    
    # Cable & input
    di_cable = (V_in_source - (i_cable * R_cable) - v_in_node) / L_cable
    di_cable = 0
    i_in_sw = i_l if high_side_on else 0.0
    v_in_node += (i_cable - i_in_sw) * (dt / C_in)
    v_sns_filt += ((i_cable * R_isns_in - v_sns_filt) / (r_filt * c_filt)) * dt
    i_cable += di_cable * dt
    i_cable = 0
    i_in_sw = i_l if high_side_on else 0.0
    
    
    
    # Soft start & output
    i_ss_down = g_m_ccl * max(0, v_sns_filt - V_in_thresh)
    v_ss = np.clip(v_ss + ((I_ss_up - i_ss_down) / C_ss) * dt, 0, 0.85)
    v_ss = max(v_ss, v_fb-0.3) # crude way to prevent soft start from "winding down" during long off times
    v_out += ((i_l - i_load) / C_out) * dt
    
    # Time advanced during this iteration
    t += dt
    # Log
    time.append(t)
    log_v_out.append(v_out)
    log_v_in_node.append(v_in_node)
    log_i_l.append(i_l)
    log_i_in_sns_avg.append(v_sns_filt/R_isns_in)
    log_v_ss.append(v_ss)
    log_i_load.append(i_load)
    log_i_cable.append(i_cable)
    log_v_comp.append(v_comp)
    log_v_c_comp.append(v_c_comp)
    log_gate.append(1.0 if high_side_on else 0.0)
    log_fan_speed.append(fan_speed)
    log_fan_current_peak.append(fan_current_peak)
    log_i_valley_target_compensated.append(i_valley_target_compensated)
    


print(f"Simulation complete: {iter_count} steps vs {int(t/200e-9):,} fixed-step iterations")
print(f"Speedup: {int(t/200e-9)/iter_count:.1f}x")

# Convert to arrays
time = np.array(time)
log_v_out = np.array(log_v_out)
log_i_l = np.array(log_i_l)
log_i_load = np.array(log_i_load)
log_v_comp = np.array(log_v_comp)
log_i_valley_target_compensated = np.array(log_i_valley_target_compensated)
log_gate = np.array(log_gate)
log_fan_speed = np.array(log_fan_speed)
log_fan_current_peak = np.array(log_fan_current_peak)
log_v_in_node = np.array(log_v_in_node)
log_dt_to_valley = np.array(log_dt_to_valley)
log_dt_to_clock = np.array(log_dt_to_clock)

# --- Plotting ---
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(10, 24), sharex=True)

ax1.plot(time*1000, log_v_out, color='tab:blue', label='V_out')
ax1.plot(time*1000, log_v_in_node/2, color='tab:green', label='V_in/2')
ax1.set_ylabel('Output Voltage (V)')
ax1.grid(True, alpha=0.3)
ax1.set_title("LM5176: Fixed Frequency Valley Current Mode (Buck)")
ax1.set_ylim([-5,45])
ax1.legend()

ax2.plot(time*1000, log_i_l, color='tab:green', marker = None, alpha=0.4, label='Inductor Current')
ax2.axhline(I_l_cbc, color='red', linestyle='--', label='CBC Peak Limit')
ax2.axhline(V_in_thresh / R_isns_in, color='grey', linestyle='--', label='Average Input Limit')
ax2.plot(time*1000, log_i_in_sns_avg, color='darkorange', label='Avg Sensed I_in')
#ax2.plot(time*1000, log_i_in_source, color='blue', label=' I_in_source', alpha = 0.3)
ax2.plot(time*1000, log_i_cable, color='blue', label=' I_cable', alpha = 0.3)
#ax2.plot(time*1000, log_i_in_filt, color='red', label='I_in_filt', alpha = 0.5)
ax2.plot(time*1000, log_i_load, color='red', label=' I_load', alpha = 0.3)
ax2.plot(time*1000, (log_v_comp - 1.6) / a_cs / R_isns_l, color='yellow', label='valley target', alpha = 0.8)
ax2.plot(time*1000, log_i_valley_target_compensated, color='purple', label='valley target compensated', alpha = 0.2)
ax2twin = ax2.twinx()
ax2twin.plot(time*1000, log_v_comp, color='lightblue', label='v_comp', alpha = 0.6)
ax2twin.plot(time*1000, log_v_c_comp, color='darkblue', label='v_c_comp', alpha =0.6)
ax2twin.plot(time*1000, log_gate/10, color='red', linestyle='--', label='mosfet gate', alpha = 0.1)
ax2.set_ylabel('Current (A)')
ax2.legend(loc='upper left')
ax2twin.legend(loc='upper right')
ax2.grid(True, alpha=0.3)
ax2.set_ylim([-35,15])
#ax2.set_xlim([30,33])

ax3.plot(time*1000, log_v_ss, color='tab:purple', label='V_ss')
ax3twin = ax3.twinx()
#ax3twin.plot(time*1000, log_i_ss_down, color='blue', label=' I_ss_down', alpha = 0.8)
#ax3twin.plot(time*1000, log_kp, color='blue', label=' kp', alpha = 0.8)
#ax3twin.plot(time*1000, log_integral, color='red', label=' ki', alpha = 0.8)
#ax3twin.plot(time*1000, log_kd, color='orange', label=' kd', alpha = 0.8)
#ax3twin.plot(time[:-1]*1000, log_dt_to_valley*1e6, color='blue', label='dt to valley (µs)', alpha = 0.8  )
#ax3twin.plot(time[:-1]*1000, log_dt_to_clock*1e6, color='green', label='dt to clock (µs)', alpha = 0.8  )
ax3twin.legend()
ax3.set_ylabel('Soft Start Voltage (V)')
#ax3twin.set_ylim([-0.2,0.2])
ax3.grid(True, alpha=0.3)


# --- continuation: fan speed / current profile panel (merged fragment) ---
# Fan speed and current profile
ax4.plot(time*1000, log_fan_speed, color='tab:blue', label='Fan Speed (Hz)')
ax4.set_ylabel('Speed (Hz)', color='tab:blue')
ax4.tick_params(axis='y', labelcolor='tab:blue')
ax4twin = ax4.twinx()
ax4twin.plot(time*1000, log_fan_current_peak, color='tab:red', linestyle='--', label='Peak Current (A)')
ax4twin.plot(time*1000, log_i_load, color='tab:red', alpha=0.5, label='Instantaneous Load (A)')
ax4twin.set_ylabel('Current (A)', color='tab:red')
ax4twin.tick_params(axis='y', labelcolor='tab:red')
ax4.set_xlabel('Time (ms)')
ax4.grid(True, alpha=0.3)
ax4.legend(loc='upper left')
ax4twin.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# --- Parameters ---
V_in_source = 20.0
# USB-C Cable Physics (1-meter typical)
L_cable = 1.0e-6      # 1uH stray inductance
R_cable = 0.15        # Resistance of wires + connectors (150mOhm)

C_in = 4*47e-6 + 2*1e-6       
V_out_nom = 12.0
I_load_max = 7.0    
L = 47e-6
C_out = 4*4700e-6 + 9 * 1e-6
r_comp = 100e3
c_comp = 220e-9 
c_hf = 150e-9
c_slope = 330e-12

# LM5176 Logic
R_isns_in = 0.1e-6 # 20e-3 
V_in_thresh = 0.050 
R_isns_l = 10e-3    
V_l_thresh = 0.080  
I_l_cbc = V_l_thresh / R_isns_l # 8.0A Peak Limit
gm_slope = 2e-6
a_cs = 5

g_m_ccl = 1e-3
v_fb_ss_offset = -18e-3
gm_ea = 1.31e-3      
C_ss = 0.1e-6
I_ss_up = 5e-6
V_ref_fb = 0.8
beta = V_ref_fb / V_out_nom
f_sw = 100000       
T_sw = 1/f_sw

# Simulation Settings
dt = 2e-7           # 100ns for cleaner switching
t_startup = 0.8
t_off = 1.6
t_total = 2.4
steps = int(t_total / dt)

# State Variables
v_in_node = 0#V_in_source
i_cable = 0
v_out = 0.0
i_l = 0.0
v_ss = 0.0
v_sns_filt = 0.0    
v_i_in_filt = 0.0
r_filt, c_filt = 1000, 10e-9
high_side_on = False
last_clock = -T_sw
# --- New State Variables for COMP Pin ---
v_comp = 1.0        # The actual voltage on the COMP pin (Top of C_hf)
v_c_comp = 1.0      # The voltage across the "bulk" capacitor C_comp only
v_fb = 0.0      # Feedback voltage
i_comp = 0.0    # Current out of Gm amplifier
cbc_limit_flag = False

# Results
time = np.linspace(0, t_total, steps)
log_v_out = np.zeros(steps)
log_v_in_node = np.zeros(steps)
log_i_l = np.zeros(steps)
log_i_in_sns_avg = np.zeros(steps)
log_v_ss = np.zeros(steps)
log_i_in_source = np.zeros(steps)
log_i_load = np.zeros(steps)
log_i_cable = np.zeros(steps)
log_i_in_filt = np.zeros(steps)
log_v_comp = np.zeros(steps)
log_v_c_comp = np.zeros(steps)
log_i_valley_target_compensated = np.zeros(steps)
log_mosfet_gate = np.zeros(steps)

for i in range(steps):
    t = time[i]
    
    # 1. Load Profile
    i_load = I_load_max if (t > t_startup and t < t_off and (t % (1/160)) < (1/320)) else 0.01
    
    # 2. FEEDBACK & ERROR AMP LOGIC (The "Pro" Way)
    # The FB pin sees V_out through the resistor divider
    v_fb = v_out * beta 
    
    # Current into COMP pin: i = gm * (Vref - Vfb)
    # Limited by the 280uA source/sink capability you provided
    i_err_amp_raw = gm_ea * (min(v_ss-v_fb_ss_offset, V_ref_fb) - v_fb)
    i_err_amp = np.clip(i_err_amp_raw, -280e-6, 280e-6)
    
    # 1. Calculate current flowing through the Resistor branch
    # This depends on the difference between the COMP pin and the bulk capacitor
    i_r_branch = (v_comp - v_c_comp) / r_comp

    # 2. Update the COMP pin voltage
    # The net current into the COMP node is (i_err_amp - i_r_branch).
    # This current charges/discharges C_hf.
    dv_comp = ((i_err_amp - i_r_branch) / c_hf) * dt
    v_comp += dv_comp

    # 3. Update the Bulk Capacitor voltage
    # The current that went through the resistor now charges C_comp.
    dv_c_comp = (i_r_branch / c_comp) * dt
    v_c_comp += dv_c_comp

    # 4. Limit v_comp to the chip's physical rails (usually 0V to ~3.3V)
    v_comp = np.clip(v_comp, 0.3, 3.0)
    
    # 4. TRANSLATE COMP TO CURRENT TARGET
    # In most controllers, I_target = (V_comp - offset) * Gain
    # Adjust 1.2 and 5.0 to match your specific IC's 
    i_valley_target = (v_comp - 1.6) / a_cs / R_isns_l

    # --- Real LM5176 Deadbeat Physics ---
    # The IC senses Vout to set the falling-edge compensation in Buck mode
    i_slope_internal = gm_slope * (v_in_node - v_out) 

    # Voltage slope on the COMP pin
    dv_slope_dt = i_slope_internal / c_slope 

    # Effective Amps slope (what the comparator 'thinks' is happening)
    se_eff = dv_slope_dt / (R_isns_l * a_cs)

    # Update the target
    t_cycle = (t - last_clock)
    slope_ramp = se_eff * t_cycle
    i_valley_target_compensated = min(I_l_cbc, i_valley_target + slope_ramp)
    log_i_valley_target_compensated[i] = i_valley_target_compensated

    
    # 3. Fixed Frequency / Valley Logic
    # Check if we are at a new clock cycle
    if i_l <= i_valley_target_compensated and not(cbc_limit_flag):
        high_side_on = True
    #if i_l >= I_l_cbc:
    ## Turn OFF if Ton is reached OR if 4A CBC Peak limit is hit
    #    cbc_limit_flag = True
    #    high_side_on = False
    #if high_side_on or cbc_limit_flag:
    #    log_i_valley_target_compensated[i] = log_i_valley_target_compensated[i-1]
    if (t - last_clock) >= T_sw:
        cbc_limit_flag = False
        if i_l>I_l_cbc:# high_side_on == False: #never turned on
            cbc_limit_flag = True
        high_side_on = False
        # In Valley Mode, we turn ON at the clock IF current is below valley setpoint
        last_clock = t
        

        
    # HS stays on for Ton. In LM5176 Buck, Ton is ~ (Vout/Vin)*T_sw
    #t_on_target = (max(v_out, 0.1) / max(v_in_node, 1.0)) * T_sw


    # 4. Inductor Physics
    v_l = (v_in_node - v_out) if high_side_on else -v_out
    i_l += (v_l / L) * dt

    # 5. USB-C CABLE & INPUT RAIL PHYSICS (The LC Pi Filter)
    # Cable current: di/dt = (V_source - V_drop_R - V_pcb) / L_cable
    di_cable = (V_in_source - (i_cable * R_cable) - v_in_node) / L_cable
    i_cable += di_cable * dt
    
    # 5. Input Rail & Average Sense
    i_in_sw = i_l if high_side_on else 0.0
    #i_in_source = (V_in_source - v_in_node)/R_source
    #v_in_node += (((V_in_source - v_in_node)/R_source) - i_in_sw) * (dt/C_in)
    v_in_node += (i_cable - i_in_sw) * (dt / C_in)
    #v_i_in_filt += ((i_in_sw * R_isns_in - v_i_in_filt) / (r_filt * c_filt)) * dt
    v_sns_filt += ((i_cable * R_isns_in - v_sns_filt) / (r_filt * c_filt)) * dt

    # 6. Soft Start & Output Cap
    i_ss_down = g_m_ccl * max(0, v_sns_filt - V_in_thresh)
    v_ss = np.clip(v_ss + ((I_ss_up - i_ss_down) / C_ss) * dt, 0, 1.21)
    v_out += ((i_l - i_load) / C_out) * dt
    
    # Log
    log_v_out[i], log_v_in_node[i], log_i_l[i], log_i_in_sns_avg[i], log_v_ss[i] = v_out, v_in_node, i_l, v_sns_filt/R_isns_in, v_ss
    #log_i_in_source[i] = (V_in_source - v_in_node)/R_cable
    log_i_load[i] = i_load
    log_i_cable[i] = i_cable
    log_i_in_filt[i] = v_i_in_filt / R_isns_in
    log_v_comp[i] = v_comp
    log_v_c_comp[i] = v_c_comp
    log_mosfet_gate[i] = high_side_on

# --- Plotting ---
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 12), sharex=True)

ax1.plot(time*1000, log_v_out, color='tab:blue', label='V_out')
ax1.plot(time*1000, log_v_in_node/2, color='tab:green', label='V_in/2')
ax1.set_ylabel('Output Voltage (V)')
ax1.grid(True, alpha=0.3)
ax1.set_title("LM5176: Fixed Frequency Valley Current Mode (Buck)")

ax2.plot(time*1000, log_i_l, color='tab:green', alpha=0.4, label='Inductor Current')
ax2.axhline(I_l_cbc, color='red', linestyle='--', label='CBC Peak Limit')
ax2.axhline(V_in_thresh / R_isns_in, color='grey', linestyle='--', label='Average Input Limit')
ax2.plot(time*1000, log_i_in_sns_avg, color='darkorange', label='Avg Sensed I_in')
#ax2.plot(time*1000, log_i_in_source, color='blue', label=' I_in_source', alpha = 0.3)
ax2.plot(time*1000, log_i_cable, color='blue', label=' I_cable', alpha = 0.3)
#ax2.plot(time*1000, log_i_in_filt, color='red', label='I_in_filt', alpha = 0.5)
ax2.plot(time*1000, log_i_load, color='red', label=' I_load', alpha = 0.3)
ax2twin = ax2.twinx()
ax2.plot(time*1000, (log_v_comp - 1.6) / a_cs / R_isns_l, color='yellow', label='valley target', alpha = 1)
ax2twin.plot(time*1000, log_v_comp, color='red', label='v_comp', alpha = 0.6)
ax2twin.plot(time*1000, log_v_c_comp, color='blue', label='v_c_comp', alpha = 0.6)
ax2twin.plot(time*1000, log_mosfet_gate/10, color='purple', label='mosfet gate', alpha = 0.6)
ax2.set_ylabel('Current (A)')
ax2.legend(loc='upper left')
ax2twin.legend(loc='upper right')
ax2.grid(True, alpha=0.3)
ax2.set_ylim([-5,30])
#ax2.set_xlim([30,33])

ax3.plot(time*1000, log_v_ss, color='tab:purple', label='V_ss')
ax3.set_ylabel('Soft Start Voltage (V)')
ax3.set_xlabel('Time (ms)')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
